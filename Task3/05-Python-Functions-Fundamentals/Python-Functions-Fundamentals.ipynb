```python
def subtract(a, b):
    """Subtract two numbers."""
    return a - b
def multiply(a, b):
    """Multiply two numbers."""
    return a * b
def divide(a, b):
    """Divide two numbers."""
    if b == 0:
        return "Cannot divide by zero."
    return a / b

print(subtract(10, 5))
print(multiply(4, 6))
print(divide(8, 2))
```

```python
def check_even_odd(num):
    """Check if a number is even or odd."""
    if num % 2 == 0:
        return "Even"
    else:
        return "Odd"

print(check_even_odd(4))
```

```python
def largest_of_three(a, b, c):
    """Return the largest of three numbers."""
    largest = a
    if b > largest:
        largest = b
    if c > largest:
        largest = c
    return largest

print(largest_of_three(10, 20, 15))
```

```python
def factorial(n):
    """Return the factorial of a number."""
    if n < 0:
        return "Factorial is not defined for negative numbers."
    elif n == 0 or n == 1:
        return 1
    else:
        result = 1
        for i in range(2, n + 1):
            result *= i
        return result
print(factorial(5))

```

```python
def is_prime(num):
    """Check if a number is prime."""
    if num <= 1:
        return False
    for i in range(2, num):
        if num % i == 0:
            return False
    return True
print(is_prime(7))
```

```python
def calculate_discount(price, discount=10):
    """Calculate the discounted price."""
    discount_amount = (discount / 100) * price
    return price - discount_amount

print(calculate_discount(100))  # Default discount of 10%
print(calculate_discount(100, 20))  # Custom discount of 20%

```

```python
def list_total(numbers):
    """Return the total of a list of numbers."""
    total = 0
    for num in numbers:
        total += num
    return total

numbers = [1, 2, 3, 4, 5]
print(list_total(numbers))
```

```python
def count_vowels(text):
    """Count the number of vowels in a string."""
    vowels = "aeiou"
    count = 0
    for char in text.lower():
        if char in vowels:
            count += 1
    return count

print(count_vowels("Hello World"))
```

```python
def is_palindrome(text):
    """Check if a string is a palindrome."""
    cleaned_text = text.lower()
    reversed_text = ""

    for char in cleaned_text:
        reversed_text = char + reversed_text
    return cleaned_text == reversed_text

print(is_palindrome("Racecar"))
print(is_palindrome("madam"))
print(is_palindrome("hello"))
```
