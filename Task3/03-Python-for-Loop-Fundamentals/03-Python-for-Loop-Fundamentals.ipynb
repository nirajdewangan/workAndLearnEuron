```python
#multiplication table of user input
n = int(input("Enter a number: "))
for i in range(1, 21):
    print(n, 'x', i, '=', n*i)
```

```python
n = int(input("Enter a number: "))
total = 0
for i in range(1, n+1):
    total += i
print("The sum of numbers from 1 to", n, "is:", total)
```

```python
number = int(input("Enter a number: "))
factorial = 1
for i in range(1, number + 1):
    factorial *= i
print("The factorial of", number, "is:", factorial)
```

```python
number = [12, 7, 9 , 15, 3, 8, 33,42]
for num in number:
    if num % 3 == 0:
        print(num)
```

```python
languages = ["Python", "Java", "C++", "JavaScript", "Ruby"]
for language in languages:
    print(language, " - ", len(language), "characters")
```

```python
student = {
    "name": "Rahul",
    "age": 22,
    "course": "Data Science",
    "city": "Bangalore"
}

for key, value in student.items():
    print(key, ":", value)
```

```python
text = input("Enter a string: ")
vowel_count = 0

for char in text.lower():
    if char in 'aeiou':
        vowel_count += 1
print("The number of vowels in the string is:", vowel_count)
```

```python
text = input("Enter a string: ")
reverse_text = ""
for char in text:
    reverse_text = char + reverse_text
print("The reverse of the string is:", reverse_text)
```

```python
number = [12, 75, 150, 180, 145, 525, 50]
largetst_number = number[0]

for num in number: 
    if num > largetst_number:
        largetst_number = num
print("The largest number in the list is:", largetst_number)
```
