```python
from neo4j import GraphDatabase

URI = "bolt://localhost:7687"
USERNAME = "neo4j"
PASSWORD = "Password12345"
driver = GraphDatabase.driver(URI, auth=(USERNAME, PASSWORD))
driver.verify_connectivity()
print("Connected to Neo4j successfully.")
```

```python
students = [
    {"name": "Amit", "age": 22},
    {"name": "Priya", "age": 25},
    {"name": "Neha", "age": 23}
]
```

```python
with driver.session() as session:
    for student in students:
        session.run(
            "CREATE (s:Student {name: $name, age: $age})",
            name=student["name"],
            age=student["age"]
        )
    print("Students created successfully.")
```

```python
courses = [
        {"name": "Python"},
    {"name": "Data Science"},
    {"name": "Generative AI"}
]
with driver.session() as session:
    for course in courses:
        session.run(
            "CREATE (c:Course {name: $name})",
            name=course["name"]
        )
    print("Courses created successfully.")
```

```python
relationships = [
    ("Amit", "Python"),
    ("Amit", "Generative AI"),
    ("Priya", "Data Science"),
    ("Neha", "Python"),
    ("Neha", "Data Science")
]
with driver.session() as session:
    for student_name, course_name in relationships:
        session.run(
            "MATCH (s:Student {name: $student_name}), (c:Course {name: $course_name}) "
            "CREATE (s)-[:ENROLLED_IN]->(c)",
            student_name=student_name,
            course_name=course_name
        )
    print("Relationships created successfully.")
```

```python
with driver.session() as session:
    result = session.run(
        "MATCH (s:Student {name:'Amit'})-[r:ENROLLED_IN]->(c:Course) "
        "RETURN s.name AS student_name, c.name AS course_name"
    )
    for record in result:
        print(f"{record['student_name']} is enrolled in {record['course_name']}")
```

```python
with driver.session() as session:
    result = session.run(
        "MATCH (s:Student) where s.age > 22 "
        "MATCH (s)-[r:ENROLLED_IN]->(c:Course) "
        "RETURN s.name AS student_name, c.name AS course_name"
    )
    for record in result:
        print(f"{record['student_name']} is enrolled in {record['course_name']}")
```
